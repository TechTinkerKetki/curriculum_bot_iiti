```python
!pip install pdfplumber camelot-py[cv]
import os
import json
from pathlib import Path
import pdfplumber
import camelot
from collections import OrderedDict
```

```python
# folders already created manually by YOU
RAW_DIR = Path("data/raw")
EXTRACTED_DIR = Path("data/processed")

PDF_PATH = RAW_DIR / "curriculum.pdf"


```

```python
text_pages = []

with pdfplumber.open(PDF_PATH) as pdf:
    for i, page in enumerate(pdf.pages):
        text = page.extract_text()
        text_pages.append({
            "page": i + 1,
            "text": text.strip() if text else ""
        })

print("Total pages extracted:", len(text_pages))
```

```python
text_out_path = EXTRACTED_DIR / "text_pages.json"

with open(text_out_path, "w", encoding="utf-8") as f:
    json.dump(text_pages, f, indent=2, ensure_ascii=False)

print("Saved text to:", text_out_path)
```

```python
tables_raw = []

tables = camelot.read_pdf(
    str(PDF_PATH),
    pages="all",
    flavor="lattice"  # works well for your PDF
)

print("Total tables detected:", tables.n)
```

```python
for idx, table in enumerate(tables):
    tables_raw.append({
        "table_id": idx,
        "page": table.page,
        "data": table.df.values.tolist()
    })
```

```python
tables_out_path = EXTRACTED_DIR / "tables_raw.json"

with open(tables_out_path, "w", encoding="utf-8") as f:
    json.dump(tables_raw, f, indent=2, ensure_ascii=False)

print("Saved tables to:", tables_out_path)
```
